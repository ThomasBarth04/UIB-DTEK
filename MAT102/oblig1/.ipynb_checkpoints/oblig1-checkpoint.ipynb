### I cellene under genereres din personlige versjon av denne oppgaven.
Kodencellene under genererer din personlige variant av denne oppgaven. Løs den genererte oppgaven som vises under den. Koden kan du ignorere.

```python
#@title Info om w(x)
md(
f'''
Kodesnutten herunder definerer en funksjon $w(x)$ som skal undersøkes i denne oppgaven. Tallet `n = {n}` er genereret ut i fra navnet ditt.
''')
```

```python
def w(x):
    res = 0
    for i in range(3):
        res = res + np.cos((n**i)*np.pi*x)/(2**i)
    return res
```

```python
#@title Instruksjoner for oppgave 1
md(
f'''
1. Plot funksjonen w(x) på intervallet $[{x0}, {x1}]$ i en liste ‘x‘ av punkter. Plottet skal ha en passende tittel, vanlig aksekors og passende xlabel og ylabel. Varier antallet av punkter i listen ‘x‘ og velg et antall der alle detaljer til grafen til funksjonen w(x)$ vises.
1. Plot den numerisk beregnede deriverte funksjonen til w(x). Gjør plottet fint som i spørsmålet over. Hva kan du si om funksjonen w(x) utifra hva du ser?
1. Lag en python funksjon `wderivert(x)` som beregner den eksakte deriverte til funksjonen w(x). Plot den eksakte deriverte og den numeriske derivete fra forrige delspørsmål i samme koordinatsystem. Lag både et plot med så få verdier i listen `x` at de to grafene er forskjellige og et plot der de to grafene er nesten like.
''')
```

```python
#1
x = np.linspace(-0.1, 0.2, 1000)
y = w(x)
plt.plot(x, y)
plt.title("w(x) på [-0.1,0.2], n = 4")
plt.xlabel('x', fontsize=12)
plt.ylabel('w(x)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color="black")
plt.axvline(x=0, color="black")
```

```python
#2
x = np.linspace(-0.1, 0.2, 1000)
y = w(x)
plt.title("w\'(x) på [-0.1,0.2], n = 4")
plt.xlabel('x', fontsize=12)
plt.ylabel('w\'(x)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.axhline(y=0, color="black")
plt.axvline(x=0, color="black")
dx = x[1] - x[0]
y_der = np.gradient(y, dx)
plt.plot(x, y_der)
plt.show()
```

Den deriverte oscillerer raskt, noe som tyder på at w(x) har mange lokale
ekstrempunkter. Amplituden til oscillasjonene i den deriverte varierer.
Den deriverte krysser null-linjen mange ganger, som bekrefter mange ekstrempunkter i w(x)


```python


#3
def wderivert(x):
    """
    w(x) = cos(n^0 * π * x) / 2^0 + cos(n^1 * π * x) / 2^1 + cos(n^2 * π * x) / 2^2
    w(x) = cos(π * x) + cos(n * π * x) / 2 + cos(n^2 * π * x) / 4

    Den deriverte blir:
    w'(x) = -π * sin(π * x) - (n * π / 2) * sin(n * π * x) - (n^2 * π / 4) * sin(n^2 * π * x)
    """

    res = 0
    for i in range(3):
        res = res - (n**i) * np.pi * np.sin((n**i) * np.pi * x) / (2**i)
    return res


fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

x_fa = np.linspace(-0.1, 0.2, 20)
w_numerisk_fa = np.gradient(w(x_fa), x_fa[1] - x_fa[0])
w_eksakt_fa = wderivert(x_fa)

axes[0].plot(x_fa, w_numerisk_fa, linewidth=2,
             markersize=6, alpha=0.7)
axes[0].plot(x_fa, w_eksakt_fa, linewidth=2,
             markersize=6, alpha=0.7)

axes[0].set_title("Få punkter (20)", fontsize=14)
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel("w'(x)", fontsize=12)
axes[0].grid(True, alpha=0.3)
axes[0].axhline(y=0, color='k', linewidth=0.5)
axes[0].axvline(x=0, color='k', linewidth=0.5)

x_mange = np.linspace(-0.1, 0.2, 1000)
w_numerisk_mange = np.gradient(w(x_mange), x_mange[1] - x_mange[0])
w_eksakt_mange = wderivert(x_mange)

axes[1].plot(x_mange, w_numerisk_mange,'r-', linewidth=2,
              alpha=0.8)
axes[1].plot(x_mange, w_eksakt_mange,'b--', linewidth=2,
              alpha=0.8)

axes[1].set_title("Mange punkter (1000)", fontsize=14)
axes[1].set_xlabel('x', fontsize=12)
axes[1].grid(True, alpha=0.3)
axes[1].axhline(y=0, color='k', linewidth=0.5)
axes[1].axvline(x=0, color='k', linewidth=0.5)

plt.suptitle("Sammenligning av numerisk og eksakt derivert", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

```

I denne oppgaven skal vi bruke numerisk integrasjon.
Vi skal bruke Simpsons regel, som tilnærmer en funksjone som en serie av andregradsfunksjoner som går gjennom endepunktene og midtpunktet på delintervaller. På figuren vises Simpsons regel for en sinusfunksjon.

Du kan lese mer om Simpson regelen (Simpson's rule på engelsk) [her](https://en.wikipedia.org/wiki/Simpson%27s_rule). Den norske Wikipedia-artikkelen har også noen tips til hvordan metoden kan implementeres, og kan leses [her](https://no.wikipedia.org/wiki/Simpson-integrasjon).

#### I cellen under genereres din personlige versjon av denne oppgaven.

```python
#@title Info om oppgave 2
md(
f'''
1. Del intervallet $[{a} , {b}]$  i to like lange delintervaller. La `y_0` være funksjonsverdien i det venstre endepunktet, la `y_1` være funksjonsverdien i midtpunktet og la `y_2` være funksjonsverdien i det høyre endepunktet av intervallet $[{a} , {b}]$. Fullfør python funksjonen `simpson_bit` under slik at den bruker Simpsons regel til å regne ut en tilnærming av integralet til $f(x) = {130 + a} \\sin(3x - \\pi/2) + {250 + b}$ på det angitte intervallet. Skriv også kommentarer i koden. (Hint: La $h$ være tallet slik at lengden til intervallet $[{a}, {b}]$ er $2h$. Formelen for Simpsons regel er da `h * (y_0 + 4*y_1 + y_2) / 3`.)
1. Gjenta punkt 1, men bruk 6, 10, og 100 delintervaller. Du kan bruke koden under hvis du vil. Husk kommentarer i koden der du beskriver hva den gjør. Hva skjer med verdien på integralet?
1. Bruk Simpsons regel på $g(x)=x^2-{1+b}x−16$ på det samme intervallet. Hva finner du ut når du prøver forskjellige antall delintervaller? Kan du forklare dette?''')
```

```python
def simpson_bit(y_0, y_1, y_2, h):
    """
    Simpsons regel på ett intervall delt i to like deler.
    """
    return h * (y_0 + 4*y_1 + y_2) / 3
```

```python
def simpson(y, h):
    if len(y) % 2 == 0:
        return 'y må bestå av et odde antall tallverdier'
    res = 0
    for i in range(((len(y) - 1) // 2)):
        y_0 = y[2*i]
        y_1 = y[2*i + 1]
        y_2 = y[2*i + 2]
        res = res + simpson_bit(y_0, y_1, y_2, h)
    return res

```

```python

#1
def f(x):
    # Definerer funksjonen f(x) = 131·sin(3x − π/2) + 256
    return 131 * np.sin(3 * x - np.pi / 2) + 256

# Venstre og høyre endepunkt for intervallet [1, 6]
a, b = 1, 6

# Intervallet deles i to like delintervaller:
h = (b - a) / 2

# venstre endepunkt
y_0 = f(a)

# midtpunktet av intervallet
y_1 = f(a + h)

# høyre endepunkt
y_2 = f(b)

# Bruker Simpsons regel til å beregne en tilnærming
integral_approx = simpson_bit(y_0, y_1, y_2, h)

# Skriver ut den tilnærmede integralverdien
print(integral_approx)
```

```python
#2
for n in [6, 10, 100]:
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = f(x)
    I = simpson(y, h)
    print(f"n = {n}, integral ≈ {I}")


'''
Observasjon:
- Verdien på integralet konvergerer raskt
- Forskjellen mellom 10 og 100 delintervaller er svært liten
- Simpsons regel har høy nøyaktighet for glatte funksjoner
'''


```
